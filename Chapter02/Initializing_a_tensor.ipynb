```python
import torch

x = torch.tensor([[1, 2]])
y = torch.tensor([[1],[2]])
```

```python
print(x.shape)
# torch.Size([1,2]) # one entity of two items
print(y.shape)
# torch.Size([2,1]) # two entities of one item each
print(x.dtype)
# torch.int64
```
