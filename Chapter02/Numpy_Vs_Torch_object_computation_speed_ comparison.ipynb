```python
import torch

x = torch.rand(1, 6400)
y = torch.rand(6400, 5000)
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
```

```python
x, y = x.to(device), y.to(device)
```

```python
%timeit z = (x @ y)
```

```python
x, y = x.cpu(), y.cpu()
%timeit z = (x @ y)
```

```python
import numpy as np

x = np.random.random((1, 6400))
y = np.random.random((6400, 5000))
%timeit z = (x @ y)
```
